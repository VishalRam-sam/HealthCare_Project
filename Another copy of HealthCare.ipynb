### Data Analysis and Data visualization

Which medical condition affects the most across the network

```python
Condition_Count = df["Medical_Condition"].value_counts()
```

```python
from matplotlib import colors
plt.figure(figsize=(8,6))
plt.bar(Condition_Count.index,Condition_Count.values,color ="#7833FE")
plt.xlabel("Medical Condition")
plt.ylabel("Numbers of Patient")
plt.xticks(rotation =90)
plt.tight_layout()
plt.title("Patients By Medical Condition")
plt.show()

```

How Has Avg Treatment Cost Changed over Time ?

```python
Avg_Cost = df["Billing_Amount"].mean()
```

```python
Avg_Cost
```

```python
df["Admission_Month"] = df['Date_Of_Admission'].dt.to_period("M")
Monthly_Avg_Cost = df.groupby("Admission_Month")["Billing_Amount"].mean()
```

```python
Monthly_Avg_Cost
```

```python
plt.figure(figsize=(15,5))
plt.plot(Monthly_Avg_Cost.index.astype(str),Monthly_Avg_Cost.values,color = "#7833FE", marker = "o", markersize = 3 )
plt.title("Average Treatment Cost Over Time (Month)")
plt.xlabel("Months")
plt.ylabel("Average Billing Amount")
plt.tight_layout()
plt.xticks(rotation = 90)
plt.show()
```

Is There a relationship between Patient Age and Treatment Cost ?

```python
Sample_df = df.sample(1000,random_state=1)
plt.figure(figsize=(7,4))
plt.scatter(Sample_df["Age"],Sample_df["Billing_Amount"],color = "red",alpha = 0.5, s=15)
plt.title("Age vs Treatment Cost")
plt.xlabel("Age")
plt.ylabel("Billing Amount")
plt.tight_layout()
plt.show()
```
